```python
#Create a list of random numbers and print all the even numbers
import random

numbers = [random.randint(1, 100) for i in range(10)]

print("List of numbers:", numbers)
print("Even numbers:")

for x in numbers:
    if x % 2 == 0:
        print(x)
```

```python
#Create a string and print it in reverse order. Now reverse the words also
text = "Bootcamp on AI"

# Reverse the complete string
print("Reverse of string:", text[::-1])

# Reverse the order of words
words = text.split()
print("Reverse of words:", " ".join(words[::-1]))
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("Libraries imported successfully! ✅")
```

```python
text = """
Mastering a dynamic lifestyle is an art that captures the beauty of human potential in time.
Whether it's the competitive rush of sports, the quiet focus of drawing and observing,
or the awe of reaching new mountains while travelling, a multidisciplinary pursuit can tell
a powerful story. Great creators use a rigorous workout or deep coding sessions as tools to
channel energy, build resilience, and design solutions for everyday life.

The core development and continuous learning of these skills, the visual framing in filming,
and overall mental stability are key elements in creating stunning personal results.
Fitness also plays a significant role in maintaining the stamina needed for navigating both
digital platforms and physically demanding terrains.

The evolution of our tools, from traditional athletics to modern software environments,
has revolutionized how we build and capture our realities.

Every achievement is a unique creation, reflecting what we believe, our individual perspective,
and our relentless drive.

I like sports fitness workout coding development filming drawing stability travelling
mountains observing learning believe Kartavya.
"""

print(text)
```

```python
words = word_tokenize(text.lower())

print(words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

print(filtered_words)
```

```python
word_freq = Counter(filtered_words)

print(word_freq)
```

```python
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate_from_frequencies(word_freq)

print("WordCloud created successfully! ✅")
```

```python
plt.figure(figsize=(10, 5))

plt.imshow(
    wordcloud,
    interpolation="bilinear"
)

plt.axis("off")
plt.title("Kartavya's Word Cloud")

plt.show()
```

```python
wordcloud.to_file("Kartavya_wordcloud.png")

print("Saved successfully as Kartavya_wordcloud.png ✅")
```

```python
import pandas as pd
import re

# Sample data
data = {
    "Name": ["  Aaradhya  ", "RAHUL", "riya!!!", "  Aman"],
    "Phone": ["98765-43210", "9876543210", "98765 43210", "98abc765432"],
    "Email": [" aaradhya@gmail.com ", "RAHUL@GMAIL.COM", "riya @gmail.com", "aman@gmail.com"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Clean Name
df["Name"] = df["Name"].str.strip()
df["Name"] = df["Name"].str.replace(r"[^A-Za-z ]", "", regex=True)

#  Clean Phone Number - keep only digits
df["Phone"] = df["Phone"].str.replace(r"\D", "", regex=True)

#  Clean Email - remove spaces and convert to lowercase
df["Email"] = df["Email"].str.replace(r"\s+", "", regex=True)
df["Email"] = df["Email"].str.lower()

print("\nCleaned Data:")
print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

time = np.linspace(0, 10, 500)
amplitude = np.sin(2 * np.pi * time)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=1000)

ax.plot(
    time,
    amplitude,
    color='crimson',
    linewidth=2,
    label="Sensor A Displacement"
)

ax.set_title(
    'Damped Harmonic Response of Structure',
    fontsize=14,
    fontweight='bold',
    pad=15
)

ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (mm)', fontsize=12)

# Enable engineering gridlines
ax.grid(True, linestyle='--', alpha=0.6)

# 
ax.legend(loc='upper right', frameon=True, shadow=True)

plt.show()
```

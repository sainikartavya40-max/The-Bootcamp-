```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("hi.jpeg")

# OpenCV BGR ko RGB me convert karna
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Kartavya Photo")
plt.axis("off")
plt.show()
```

```python
img_resize = cv2.resize(img, (400, 256))

img_resize_rgb = cv2.cvtColor(img_resize, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 5))
plt.imshow(img_resize_rgb)
plt.title("Kartavya Photo - Resized")
plt.axis("off")
plt.show()
```

```python
img_flip = cv2.flip(img, 0)

img_flip_rgb = cv2.cvtColor(img_flip, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 5))
plt.imshow(img_flip_rgb)
plt.title("Kartavya Photo - Flipped")
plt.axis("off")
plt.show()
```

```python
img_crop = img[100:300, 200:500]

img_crop_rgb = cv2.cvtColor(img_crop, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 5))
plt.imshow(img_crop_rgb)
plt.title("Kartavya Photo - Cropped")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("Kartavya_resized.jpeg", img_resize)

print("Resized image saved successfully!")
```

```python
cv2.imwrite("Kartavya_flipped.jpeg", img_flip)

print("Flipped image saved successfully!")
```

```python
cv2.imwrite("Kartavya_cropped.jpeg", img_crop)

print("Cropped image saved successfully!")
```

```python
import cv2
from ultralytics import settings
from ultralytics import YOLO
import matplotlib.pyplot as plt

# Disable data and crash report tracking
settings.update({"sync": False})

# Load YOLOv8 Nano model
model = YOLO("yolov8n.pt")

# Read Kartavya's image
img = cv2.imread("hi.jpeg")

if img is None:
    print("Error: hi.jpeg not found!")
else:
    # Perform object detection
    results = model(img, verbose=False)

    # Count detected objects
    detected_count = len(results[0].boxes)

    # Draw detection boxes
    annotated_frame = results[0].plot()

    # Add object count
    cv2.putText(
        annotated_frame,
        f"Total Objects Counted: {detected_count}",
        (20, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2,
        cv2.LINE_AA
    )

    # Convert BGR to RGB
    annotated_rgb = cv2.cvtColor(
        annotated_frame,
        cv2.COLOR_BGR2RGB
    )

    # Display inside Jupyter
    plt.figure(figsize=(12, 8))
    plt.imshow(annotated_rgb)
    plt.title("Kartavya AI Demo - Object Detection")
    plt.axis("off")
    plt.show()

    print("Total Objects Detected:", detected_count)
```

```python
import os
import sys
import time
import cv2

# Load OpenCV's face detector directly from the current folder
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

if face_cascade.empty():
    raise RuntimeError('Failed to load the face detection model. Ensure the XML file is in the same folder.')

# Using 1 to bypass the Mac Continuity Camera (iPhone) if needed
cap = cv2.VideoCapture(1)
if not cap.isOpened():
    raise RuntimeError('Could not open the webcam.')

# On macOS, a desktop app is needed to show the OpenCV window.
# In headless environments (like some terminal-only sessions), skip imshow.
show_window = os.environ.get('DISPLAY') is not None or sys.platform == 'darwin'
window_name = 'BCA AI Demo - Face Detection'

try:
    while True:
        success, frame = cap.read()
        if not success:
            print('Failed to read webcam frame.')
            break

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(
            gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30)
        )

        for x, y, w, h in faces:
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        if show_window:
            cv2.imshow(window_name, frame)
            if cv2.waitKey(1) & 0xFF == ord('q'):
                break
        else:
            print(f'Faces detected: {len(faces)}', end='\r')
            time.sleep(0.1)

except Exception as e:
    print(f'Runtime error: {e}')
    print('If you are running in a terminal-only environment, open this script in a desktop terminal or enable camera access.')
finally:
    cap.release()
    if show_window:
        cv2.destroyAllWindows()
```

```python
import numpy as np

# 1D dataset: hours spent on hobbies per week
hobby_hours = np.array([3, 5, 4, 6, 5, 3, 7, 5, 4, 6])

# Mean
mean = np.mean(hobby_hours)

# Median
median = np.median(hobby_hours)

# Mode
values, counts = np.unique(hobby_hours, return_counts=True)
mode = values[np.argmax(counts)]

# Standard Deviation
std_deviation = np.std(hobby_hours)

# Display results
print("Hobby Hours Dataset:", hobby_hours)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Standard Deviation:", std_deviation)
```

```python
import numpy as np

# 2D Dataset
# Columns: Dancing, Painting, Singing, Reading, Gaming, Photography

hobbies = np.array([
    [3, 2, 3, 5, 6, 2],
    [5, 4, 4, 6, 8, 3],
    [4, 3, 5, 4, 7, 4],
    [6, 5, 3, 7, 5, 2],
    [5, 4, 6, 5, 9, 5],
    [3, 2, 4, 6, 6, 3],
    [7, 6, 5, 3, 8, 4],
    [5, 4, 3, 5, 7, 2],
    [4, 3, 4, 4, 6, 3],
    [6, 5, 5, 6, 10, 4]
])

hobby_names = [
    "Dancing",
    "Painting",
    "Singing",
    "Reading",
    "Gaming",
    "Photography"
]

# Calculate statistics for each hobby
for i in range(len(hobby_names)):

    data = hobbies[:, i]

    # Mode
    values, counts = np.unique(data, return_counts=True)
    mode = values[np.argmax(counts)]

    print("\nHobby:", hobby_names[i])
    print("Mean:", np.mean(data))
    print("Median:", np.median(data))
    print("Mode:", mode)
    print("Standard Deviation:", np.std(data))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate engineering data using NumPy
time = np.linspace(0, 10, 500)  # 500 points from 0 to 10 seconds
amplitude1 = np.sin(2 * np.pi * time)  # Simple sine wave
amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)  # Damped sine wave

# 2. Create a figure with 2 subplots vertically (2 rows, 1 column)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), dpi=100)

# 3. Plot 1: Simple Sine Wave (Top Subplot)
ax1.plot(
    time,
    amplitude1,
    color="crimson",
    linewidth=2,
    label="Undamped Harmonic Motion",
)
ax1.set_title(
    "Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax1.set_xlabel("Time (seconds)", fontsize=11)
ax1.set_ylabel("Displacement (mm)", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right", frameon=True, shadow=True)

# 4. Plot 2: Damped Sine Wave (Bottom Subplot)
ax2.plot(
    time, amplitude2, color="navy", linewidth=2, label="Damped Harmonic Motion"
)
ax2.set_title(
    "Damped Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax2.set_xlabel("Time (seconds)", fontsize=11)
ax2.set_ylabel("Displacement (mm)", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right", frameon=True, shadow=True)

# 5. Adjust layout to avoid overlapping labels and display
plt.tight_layout()
# plt.savefig('sensor_data_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
